```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import mean_squared_error
```

```python
dataset = pd.read_csv('Merged_Resistivity_Hardness_Data.csv')
X = dataset.iloc[:, 0:13].values
y = dataset["Hardness"].values
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size = 1/3, random_state = 0
)

from sklearn.preprocessing import StandardScaler

sc_X = StandardScaler()
X_train = sc_X.fit_transform(X_train)
X_test = sc_X.transform(X_test)

sc_y = StandardScaler()
y_train = sc_y.fit_transform(y_train)

from sklearn.linear_model import LinearRegression
regressor = LinearRegression()

regressor.fit(X_train, y_train)

print(regressor.coef_)
print(regressor.intercept_)

y_pred = regressor.predict(X_test)

print(y_test.shape)
print(y_pred.shape)

from sklearn.metrics import r2_score

y_pred_inv = sc_y.inverse_transform(y_pred)
y_test_inv = y_test.flatten()

result = r2_score(y_test_inv, y_pred_inv)
print(result)
```

```python
plt.scatter(X_train[:, 0], y_train.flatten(), color = 'red')
plt.plot(X_train[:, 0], regressor.predict(X_train).flatten(), color = 'blue')

plt.title('Hardness')
plt.xlabel('')
plt.ylabel('Hardness (Scaled)')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import torch

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

dataset = pd.read_csv('Merged_Resistivity_Hardness_Data.csv')
X = dataset.iloc[:, 0:13].values
y = dataset["Hardness"].values

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 1/3, random_state = 0)

# -----------------------------
# w1을 스칼라에서 13차원 벡터로 변경
# -----------------------------
w1 = np.ones(13) * 10       # [10,10,10,...]
w0 = 10
lr = 0.0025

def forward(x):
    return np.sum(w1 * x) + w0 ##차원의 증가로 기존의 식 대체

def loss_fn(y_pred, y):
  return (y_pred - y) * (y_pred - y)

def update(x, y, y_pred):
	global w1, w0
	w1 = w1 - lr * 2 * x * (y_pred - y)
	w0 = w0 - lr * 2 * 1 * (y_pred - y)
print(w1, w0)

for epoch in range(100):
	for x_val, y_val in zip(X_train, y_train):
		y_pred = forward(x_val)
		loss = loss_fn(y_pred, y_val)
		update(x_val, y_val, y_pred)

print(w1, w0)

predict = np.array([(np.dot(w1, x) + w0) for x in X_test])

u = np.sum((predict - y_test)**2)
v = np.sum((y_test - np.average(y_test))**2)

result = 1 - u/v
print(result)

from sklearn.metrics import r2_score

result = r2_score(y_test, predict)

print(result)
```
